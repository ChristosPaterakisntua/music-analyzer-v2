### Parsed Dataset Compression

Compress the thousand .npz files to a .tar file.

Nessecary because i/o is very slow for google colab (took me ~2-3 h) so we import just one file and then decompress it. The time is reduced to just ~2-3 min.

This process might take some time but it will only be once and will be cpu bound, saving gpu time

```python
import shutil
import subprocess
from pathlib import Path
from google.colab import drive
```

```python
drive.mount("/content/drive")
```

```python
drive_parsed = Path("/content/drive/MyDrive/music-analyzer_v2/parsed_dataset")
local_tar = Path("/content/parsed_dataset.tar")
drive_tar = Path("/content/drive/MyDrive/music-analyzer_v2/parsed_dataset.tar")
```

```python
subprocess.run(
    ["tar", "-cf", str(local_tar), "-C", str(drive_parsed), "."],
    check=True,
)
```

```python
shutil.copyfile(local_tar, drive_tar)
local_tar.unlink()

print(f"Ready: {drive_tar} ({drive_tar.stat().st_size / (1024**3):.2f} GiB)")
```
